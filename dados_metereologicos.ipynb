```python
import pandas as pd
import requests
import time

endpoint_planilha = 'https://docs.google.com/spreadsheets/d/1nduw1me1uVmsGkamriylKnPNA4wlB0O52vEtMgJe_oM/export?format=csv'
endpoint_api = 'https://api.open-meteo.com/v1/forecast'
```

```python
print('--- Dados Coletados ---')
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
# Iteração de cada resposta do formulário
for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do local']
  lat = linha['Latitude']
  lon = linha['Longitude']

  parametros = {
      'latitude': lat,
      'longitude': lon,
      'current': 'temperature_2m'
  }

  print(f'---- {nome_local} ----')
  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print('\n\n')
```
